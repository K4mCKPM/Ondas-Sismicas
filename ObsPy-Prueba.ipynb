```python
from obspy.clients.fdsn import Client
from obspy.clients.fdsn.header import URL_MAPPINGS

SGC = "https://sismo.sgc.gov.co:8443"
client = Client(SGC)
```

```python
from obspy import UTCDateTime
t = UTCDateTime("2026-02-27T06:45:00.000")
st = client.get_waveforms("CM", "BOG", "*", "*", t, t + 60 * 60)
st.plot()
```

```python

import cartopy
import matplotlib.pyplot as plt
starttime = UTCDateTime("1979-06-22")
endtime = UTCDateTime("2026-10-23")
inventory = client.get_stations(network="CM", station="*",
                                starttime=starttime,
                                endtime=endtime)
print(inventory)

```

```python
inventory.plot(projection="local", show="False")
```
