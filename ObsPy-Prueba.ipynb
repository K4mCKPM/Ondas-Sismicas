```python
from obspy.clients.fdsn import Client

SGC = "https://sismo.sgc.gov.co:8443"   #URL del Servicio Geologico Colombiano
client = Client(SGC)
```

```python
from obspy import UTCDateTime
t = UTCDateTime("2026-08-10T12:30:00.000") # Sismo 10 de agosto (UTC 12:34:27)
st = client.get_waveforms("CM", "BOG", "*", "*", t, t + 60 * 30) #Se grafica media hora después del tiempo inicial t
st.plot()
```

```python

import cartopy
import matplotlib.pyplot as plt
starttime = UTCDateTime("2016-09-12")
endtime = UTCDateTime("2026-11-16")
inventory = client.get_stations(network="CM", station="*",
                                starttime=starttime,
                                endtime=endtime)
print(inventory)

```

```python
inventory.plot(projection="local", show="False")
```
